Practical 3: TF-IDF Vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
import pandas as pd
```

Create a Corpus

```python
corpus = [

"machine learning transform artificial intelligence",

"machine learning use neural network",

"deep learning subset machine learning"

]
```

```python
print("Corpus")
print("-"*50)

for i, doc in enumerate(corpus,1):
    print(f"Document {i}: {doc}")
```

Create TF-IDF Vectorizer

```python
vectorizer = TfidfVectorizer()
```

```python
vectorizer.fit(corpus)
```

Learn Vocabulary

```python
print(vectorizer.vocabulary_)
```


Print Feature Names

```python
features = vectorizer.get_feature_names_out()

print(features)
```

Convert Documents into TF-IDF Matrix

```python
tfidf_matrix = vectorizer.transform(corpus)
```

Convert Matrix into Array

```python
matrix = tfidf_matrix.toarray()

print(matrix)
```

Convert into DataFrame

```python
df = pd.DataFrame(

matrix,

columns=features

)

print(df)
```

TF-IDF of One Document

```python
print(df.iloc[0])
```

TF-IDF of One Word

```python
print(df["machine"])
```

Save the Matrix


```python
df.to_csv("tfidf_matrix.csv",index=False)

print("TF-IDF Matrix Saved Successfully")
```
